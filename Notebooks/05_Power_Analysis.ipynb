# 1. Power Analysis

## Objective

Previous notebooks established whether the observed differences in player retention were statistically significant.

This notebook evaluates the quality of the experiment itself by answering the following questions:

- Did the experiment include enough players?
- Was the sample size sufficient to detect the observed effects?
- Could a meaningful difference have gone unnoticed?
- How should similar experiments be designed in the future?

Power analysis complements hypothesis testing by assessing the ability of an experiment to detect true treatment effects.

# 2. Why Power Analysis Matters

## Statistical significance is only part of the story.

A statistically significant result does not necessarily imply that the experiment was well designed.

Power analysis helps determine whether the experiment had a high probability of detecting a true effect if one actually existed.

Without adequate statistical power:

- Real improvements may go undetected.
- Business decisions may be based on inconclusive evidence.
- Additional experimentation may be required.

# 3. Statistical Power

## 3.1 Objective

Estimate the statistical power of the Cookie Cats experiment using the observed effect size and sample sizes.

Statistical power represents the probability of correctly detecting a true difference between the experiment groups when one actually exists.

## 3.2 Theory

Statistical power is the probability of rejecting the null hypothesis when a true treatment effect exists.

Power depends on four key factors:

- Sample size
- Effect size
- Significance level (α)
- Data variability

A commonly accepted benchmark is **80% power**, meaning the experiment has an 80% chance of detecting a meaningful effect if it truly exists.

Higher statistical power reduces the likelihood of false negatives and improves confidence in experimental conclusions.

## 3.3 Compute Statistical Power

Estimate the statistical power using the observed effect size from the seven-day retention analysis and the sample sizes of both experiment groups.

The calculation assumes a two-sided hypothesis test with a significance level of 5%.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from scipy.stats import chisquare
```

```python
df = pd.read_csv("../data/cleaned/cookie_cats_cleaned.csv")
```

```python
from statsmodels.stats.power import NormalIndPower

analysis = NormalIndPower()

# Observed Cohen's h from Notebook 03
effect_size = 0.0211

# Smaller group size
nobs = min(44700, 45489)

alpha = 0.05

power = analysis.solve_power(
    effect_size=effect_size,
    nobs1=nobs,
    alpha=alpha,
    ratio=1,
    alternative="two-sided"
)

print(f"Observed Effect Size : {effect_size:.4f}")
print(f"Sample Size          : {nobs}")
print(f"Significance Level   : {alpha}")
print(f"Statistical Power    : {power:.4f}")
```

## 3.4 Interpretation

### Results

- **Observed Effect Size (Cohen's h):** 0.0211
- **Sample Size:** 44,700 players per group (minimum group size)
- **Statistical Power:** 88.39%

### Key Insights

- The experiment achieved **88.39% statistical power**, exceeding the commonly accepted benchmark of 80%.
- The available sample size was sufficient to detect the observed retention difference with high probability.
- The likelihood of failing to detect a true effect (Type II Error) was relatively low.

### Business Perspective

The Cookie Cats experiment was adequately powered, indicating that the observed seven-day retention improvement is unlikely to be the result of an insufficient sample size. This increases confidence in the reliability of the experimental findings.

# 4. Minimum Detectable Effect (MDE)

## 4.1 Objective

Estimate the smallest retention improvement that this experiment could reliably detect given its sample size, significance level, and desired statistical power.

## 4.2 Theory

The **Minimum Detectable Effect (MDE)** is the smallest effect size that an experiment can detect with a specified level of confidence and statistical power.

MDE depends on:

- Sample size
- Statistical power
- Significance level (α)

Experiments with larger sample sizes can reliably detect smaller effects, while smaller experiments require larger differences before statistical significance can be achieved.

MDE is commonly used when planning A/B tests to determine the required sample size before launching an experiment.

## 4.3 Compute the Minimum Detectable Effect

Estimate the minimum standardized effect size that this experiment could detect while maintaining 80% statistical power at a 5% significance level.

```python
desired_power = 0.80

mde = analysis.solve_power(
    power=desired_power,
    nobs1=nobs,
    alpha=alpha,
    ratio=1,
    alternative="two-sided"
)

print(f"Desired Statistical Power : {desired_power:.2f}")
print(f"Minimum Detectable Effect (Cohen's h): {mde:.4f}")
```

## 4.4 Interpretation

### Results

- **Desired Statistical Power:** 80%
- **Minimum Detectable Effect (Cohen's h):** 0.0187

### Key Insights

- The experiment could reliably detect standardized effects of **0.0187 or larger**.
- The observed effect size (**0.0211**) exceeded the minimum detectable effect.
- This indicates that the experiment had sufficient sensitivity to identify the observed difference in seven-day retention.

### Business Perspective

The experiment was capable of detecting very small changes in player retention. Since the observed effect exceeded the minimum detectable threshold, the statistical significance observed in the seven-day analysis is supported by an appropriately sensitive experimental design.

# 5. Sample Size Analysis

## 5.1 Objective

Estimate the minimum number of players required in each experiment group to detect the observed effect size with 80% statistical power at a 5% significance level.

## 5.2 Compute Required Sample Size

Estimate the required sample size per experiment group using the observed effect size from the seven-day retention analysis.

```python
required_sample = analysis.solve_power(
    effect_size=effect_size,
    power=0.80,
    alpha=0.05,
    ratio=1,
    alternative="two-sided"
)

print(f"Required Sample Size per Group : {required_sample:.0f}")
print(f"Actual Sample Size per Group   : {nobs}")
```

## 5.3 Interpretation

### Results

- **Required Sample Size per Group:** 35,259 players
- **Actual Sample Size per Group:** 44,700 players

### Key Insights

- The actual sample size exceeded the required sample size by approximately **9,441 players per group**.
- The experiment included enough participants to detect the observed effect with the desired statistical power.
- The available sample size increased the reliability and precision of the statistical estimates.

### Business Perspective

The Cookie Cats experiment was designed with a sufficiently large sample size to detect small but meaningful changes in player retention. This supports the credibility of the experimental conclusions and suggests that future experiments targeting similar effect sizes should aim for a comparable sample size.
