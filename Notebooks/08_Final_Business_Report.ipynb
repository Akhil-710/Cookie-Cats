# 1. Executive Summary

## Objective

This report presents the findings of an A/B experiment conducted on the mobile game **Cookie Cats** to evaluate whether moving the first progression gate from **Level 30** to **Level 40** affects player retention.

The analysis combines exploratory data analysis, statistical hypothesis testing, bootstrap validation, statistical power analysis, and player segmentation to assess the impact of the experimental change.

The objective is to provide evidence-based recommendations that support product decision-making and improve long-term player engagement.

# 2. Business Problem

Progression gates are commonly used in free-to-play games to regulate player progression, encourage engagement, and create opportunities for monetization.

King conducted an experiment to determine whether delaying the first progression gate from Level 30 to Level 40 would improve the player experience and increase retention.

The primary business question addressed in this project is:

> **Does moving the progression gate from Level 30 to Level 40 improve player retention, and should the product team adopt this change?**

# 3. Dataset Overview

The dataset contains player-level information collected during the A/B experiment.

### Dataset Summary

- Total observations: **90,189 players**
- Experiment groups:
  - Gate 30: **44,700 players**
  - Gate 40: **45,489 players**

### Features

| Feature | Description |
|----------|-------------|
| userid | Unique player identifier |
| version | Experiment group assignment |
| sum_gamerounds | Total number of game rounds played |
| retention_1 | Returned one day after installation |
| retention_7 | Returned seven days after installation |

The dataset provides sufficient sample size for reliable statistical analysis.

# 4. Analytical Methodology

The project followed a structured analytical workflow consisting of multiple stages.

1. Data understanding and validation
2. Exploratory data analysis (EDA)
3. Sample Ratio Mismatch (SRM) verification
4. Hypothesis testing using two-proportion Z-tests
5. Confidence interval estimation
6. Effect size measurement using Cohen's h
7. Bootstrap validation
8. Statistical power analysis
9. Player engagement segmentation
10. Executive dashboard creation

Each stage was designed to validate the experimental results from a different analytical perspective, ensuring robust and reliable conclusions.

# 5. Key Findings

The analysis produced several important insights regarding player retention and the impact of changing the first progression gate.

## 5.1 Exploratory Data Analysis

- The player engagement distribution was highly right-skewed, indicating that most players completed relatively few game rounds while a small group of highly engaged players contributed disproportionately high activity.
- One extreme outlier (49,854 game rounds) was identified. Removing this observation reduced the skewness substantially without changing the overall conclusions.
- No missing values or duplicate records were detected, indicating high data quality.

---

## 5.2 Sample Ratio Mismatch (SRM)

The experiment groups contained 44,700 and 45,489 players respectively.

Although a statistically significant SRM was detected, the imbalance represented less than 1% of the total sample and was considered sufficiently small for the subsequent analyses to remain reliable.

---

## 5.3 One-Day Retention

The comparison between the two experiment groups showed:

- Gate 30 Retention: **44.82%**
- Gate 40 Retention: **44.23%**
- Absolute Difference: **0.59 percentage points**

The two-proportion Z-test produced a **p-value of 0.0744**, indicating that the observed difference was **not statistically significant** at the 5% significance level.

---

## 5.4 Seven-Day Retention

The seven-day retention analysis showed:

- Gate 30 Retention: **19.02%**
- Gate 40 Retention: **18.20%**
- Absolute Difference: **0.82 percentage points**

The Z-test produced a **p-value of 0.0016**, demonstrating a statistically significant improvement in long-term retention for players assigned to Gate 30.

---

## 5.5 Validation Analyses

Bootstrap resampling produced confidence intervals that closely matched the analytical confidence intervals obtained through hypothesis testing.

Statistical power analysis showed an achieved power of **88.39%**, exceeding the commonly recommended threshold of 80%.

These results strengthen confidence in the validity of the experimental findings.

---

## 5.6 Player Segmentation

Segment-wise analysis revealed that:

- Low engagement players showed no statistically significant difference.
- Medium engagement players experienced significantly higher retention under Gate 30.
- High engagement players also demonstrated significantly higher retention under Gate 30.

This suggests that the overall improvement in long-term retention is primarily driven by players who engage more deeply with the game.

# 6. Business Recommendations

Based on the statistical evidence generated throughout this analysis, the following recommendations are proposed.

## Recommendation 1

Maintain the first progression gate at **Level 30**.

The experiment consistently demonstrates higher long-term player retention for Gate 30 while providing no meaningful improvement from moving the gate to Level 40.

---

## Recommendation 2

Continue monitoring long-term retention rather than focusing solely on one-day retention.

Although one-day retention remained similar between the two variants, meaningful differences emerged after seven days, making long-term metrics more valuable for product evaluation.

---

## Recommendation 3

Prioritize highly engaged players during future experimentation.

The segmentation analysis indicates that Medium and High engagement players were most affected by the gate placement.

Optimizing the experience for these player groups is likely to produce greater improvements in long-term retention and lifetime value.

# 7. Limitations

Although the analysis provides strong statistical evidence, several limitations should be considered.

- The dataset contains only player retention outcomes and does not include monetization metrics.
- Player demographics and acquisition channels were unavailable.
- The experiment evaluates only one progression gate placement (Level 30 versus Level 40).
- The analysis assumes that players were randomly assigned to experiment groups.
- External factors such as seasonality or concurrent game updates were not available for evaluation.

These limitations should be considered when generalizing the findings to future experiments.

# 8. Future Work

Several opportunities exist for extending this analysis.

- Evaluate additional progression gate locations beyond Levels 30 and 40.
- Incorporate monetization metrics such as in-app purchases and ad revenue.
- Analyze player lifetime value (LTV) alongside retention.
- Perform cohort-based retention analysis across different acquisition periods.
- Apply survival analysis techniques to model player churn over time.
- Develop predictive machine learning models to identify players at high risk of churn.

These analyses would provide a more comprehensive understanding of player behavior and support future product decisions.

# 9. Final Conclusion

This project evaluated the impact of changing the first progression gate in Cookie Cats from **Level 30** to **Level 40** using a comprehensive statistical analysis pipeline.

Across exploratory data analysis, hypothesis testing, bootstrap validation, statistical power analysis, and player segmentation, the evidence consistently supports maintaining the progression gate at **Level 30**.

While no statistically significant difference was observed in one-day retention, Gate 30 produced a significant improvement in seven-day retention, particularly among Medium and High engagement players.

These findings suggest that maintaining the original progression gate maximizes long-term player engagement without negatively affecting early player retention.

This project demonstrates how data-driven experimentation and statistical inference can support evidence-based product decisions in the gaming industry.
